```python
%%writefile app.py
import os
import re
from langchain_community.document_loaders import PDFPlumberLoader, TextLoader
from langchain_core.documents import Document
import streamlit as st
from langchain_ollama import ChatOllama
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser

st.set_page_config(page_title="المساعد الأكاديمي لجامعة حلوان الأهلية", page_icon="🎓")
st.title("🎓 المساعد الأكاديمي (برنامج هندسة النظم الذكية)")

# --- 👤 بيانات الطالب ---
with st.sidebar:
    st.header("👤 بيانات الطالب")
    if "student_name" not in st.session_state:
        st.session_state.student_name = ""
    if "student_level" not in st.session_state:
        st.session_state.student_level = "غير محدد"

    st.session_state.student_name = st.text_input("اسمك إيه؟", value=st.session_state.student_name)
    st.session_state.student_level = st.selectbox(
        "المستوى الدراسي:",
        ["غير محدد", "المستوى الأول", "المستوى الثاني", "المستوى الثالث", "المستوى الرابع", "المستوى الخامس"],
        index=["غير محدد", "المستوى الأول", "المستوى الثاني", "المستوى الثالث", "المستوى الرابع", "المستوى الخامس"].index(st.session_state.student_level)
    )
    st.markdown("---")
    st.info("💡 حدد مستواك عشان المساعد يقدر يجاوبك بدقة في تسجيل المواد.")
# ---------------------------------------------------

def render_message(text):
    arabic_chars = len(re.findall(r'[\u0600-\u06FF]', text))
    english_chars = len(re.findall(r'[a-zA-Z]', text))

    if arabic_chars > english_chars:
        st.markdown(f'<div dir="rtl" style="text-align: right;">\n\n{text}\n\n</div>', unsafe_allow_html=True)
    else:
        st.markdown(f'<div dir="ltr" style="text-align: left;">\n\n{text}\n\n</div>', unsafe_allow_html=True)

@st.cache_resource
def init_rag_system():
    docs = []

    # 1. قراءة اللائحة (PDF) فقط للبحث عن المواد
    if os.path.exists("/content/rules_2.PDF"):
        loader = PDFPlumberLoader("/content/rules_2.PDF")
        docs.extend(loader.load())
    elif os.path.exists("/content/rules.PDF"):
        loader = PDFPlumberLoader("/content/rules.PDF")
        docs.extend(loader.load())

    # 2. خريطة المبنى مدمجة كـ Document
    building_map_text = """
    دليل أماكن وقاعات ومدرجات ومعامل جامعة حلوان الأهلية - مبنى هندسة (1):
    - الدور الأرضي: قاعة المجسمات، فصل (1)، فصل (2)، فصل (3)، فصل (4)، فصل (5)، فصل (6)، فصل (7).
    - الدور الأول: فصل (8)، فصل (9)، فصل (10)، مرسم (1)، مرسم (2)، صالة رسم (1)، صالة رسم (2)، معمل كمبيوتر (1).
    - الدور الثاني: فصل (11)، فصل (14)، فصل (15)، فصل (16)، فصل (17)، فصل (18)، فصل (19)، فصل (20)، مدرج (1)، مدرج (2)، معمل كمبيوتر (2)، معمل كمبيوتر (3)، معمل كمبيوتر (4) (ملاحظة: يستخدم حاليا كـ كنترول فنون).
    - الدور الثالث: مدرج (3)، مدرج (4)، معمل كمبيوتر (5)، معمل كمبيوتر (6).
    """
    docs.append(Document(page_content=building_map_text))

    # 3. قراءة المعلومات العامة للشروط
    if os.path.exists("/content/general_info.txt"):
        loader = TextLoader("/content/general_info.txt", encoding="utf-8")
        docs.extend(loader.load())

    if not docs:
        return None, None, "لم يتم العثور على ملف اللائحة!"

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1500,
        chunk_overlap=400,
        separators=["\n\n", "\n", " "]
    )
    chunks = text_splitter.split_documents(docs)

    my_gemini_key = ""YOUR_GEMINI_API_KEY_HERE""

    embeddings = GoogleGenerativeAIEmbeddings(
        model="gemini-embedding-001",
        google_api_key=my_gemini_key
    )

    vector_store = FAISS.from_documents(chunks, embeddings)
    retriever = vector_store.as_retriever(search_kwargs={"k": 8})
    llm = ChatOllama(model="qwen2.5:7b", temperature=0.1, num_predict=1024)

    return retriever, llm, "جاهز للرد!"

with st.spinner("جاري تحميل النظام وتحليل اللائحة..."):
    retriever, llm, status_message = init_rag_system()

if retriever is None:
    st.error(status_message)
    st.stop()

if st.button("🗑️ مسح المحادثة"):
    st.session_state.messages = []
    st.rerun()

if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        render_message(msg["content"])

if len(st.session_state.messages) == 0:
    st.markdown("<h4 style='text-align: right; direction: rtl;'>💡 أسئلة شائعة يمكنك البدء بها:</h4>", unsafe_allow_html=True)
    col1, col2, col3 = st.columns(3)

    with col3:
        if st.button("🗺️ أين تقع القاعات والمدرجات؟"):
            st.session_state.quick_query = "أين يقع مدرج 1 أو قاعات السكاشن؟"
    with col2:
        if st.button("📚 ما هي مواد المستوى الأول؟"):
            st.session_state.quick_query = "ما هي مواد المستوى الأول وما متطلباتها؟"
    with col1:
        if st.button("🎓 ما هي شروط التخرج؟"):
            st.session_state.quick_query = "ما هي شروط التخرج من برنامج هندسة النظم الذكية؟"

user_input = st.chat_input("اسأل عن أي مستوى، مادة، متطلب، أو مكان قاعة...")

if 'quick_query' in st.session_state:
    user_input = st.session_state.quick_query
    del st.session_state.quick_query

if user_input:
    st.session_state.messages.append({"role": "user", "content": user_input})
    with st.chat_message("user"):
        render_message(user_input)

    with st.chat_message("assistant"):
        with st.spinner("جاري البحث..."):
            try:
                is_arabic = bool(re.search(r'[\u0600-\u06FF]', user_input))

                chat_history = ""
                for msg in st.session_state.messages[-5:-1]:
                    sender = "الطالب" if msg["role"] == "user" else "المساعد"
                    chat_history += f"{sender}: {msg['content']}\n"

                last_user_msgs = [m["content"] for m in st.session_state.messages[-5:-1] if m["role"] == "user"]
                if len(user_input.split()) <= 5 and last_user_msgs:
                    search_query = f"{last_user_msgs[-1]} {user_input}"
                else:
                    search_query = user_input

                name = st.session_state.student_name if st.session_state.student_name else "طالب"
                level = st.session_state.student_level

                if level == "غير محدد":
                    student_context = f"مستوى الطالب {name} الحالي: (غير محدد)."
                else:
                    student_context = f"مستوى الطالب {name} الحالي: ({level})."

                # معلومات لا تقبل الخطأ محقونة مباشرة في عقل الموديل
                hardcoded_knowledge = """
--- معلومات ثابتة (يجب الاعتماد عليها دائماً) ---
1. شروط التخرج: اجتياز 148 ساعة معتمدة بنجاح.
2. مشروع التخرج: دراسة مشروع تخرج (1) و (2) تتم في المستوى الثالث.
3. دليل المباني والأماكن (مبنى هندسة 1):
   - قاعة المجسمات والفصول من 1 إلى 7: تقع في الدور الأرضي.
   - الفصول (8، 9، 10): تقع في الدور الأول.
   - مرسم 1 و 2، وصالة رسم 1 و 2: تقع في الدور الأول.
   - معمل كمبيوتر 1: يقع في الدور الأول.
   - الفصول (11، ومن 14 إلى 20): تقع في الدور الثاني.
   - مدرج 1: يقع في الدور الثاني.
   - مدرج 2: يقع في الدور الثاني.
   - معمل كمبيوتر 2 و 3 و 4: تقع في الدور الثاني.
   - مدرج 3: يقع في الدور الثالث.
   - مدرج 4: يقع في الدور الثالث.
   - معمل كمبيوتر 5 و 6: تقع في الدور الثالث.
--------------------------------------------------
"""

                if is_arabic:
                    system_template = f"""أنت مرشد أكاديمي ذكي ولبق لبرنامج هندسة النظم الذكية.
{student_context}

تعليمات الرفض الذكي (التزم بالحدود ولكن بأسلوب طبيعي ومفهوم):
1. الأسئلة الخارجية: إذا سأل الطالب عن رياضة، طبخ، أو معلومات عامة، تفاعل مع سؤاله بلطف في نصف جملة لتبين أنك فهمته، ثم اعتذر بمهنية ووضح أن تخصصك هو الإرشاد الأكاديمي للكلية فقط. (مثال: "سؤالك عن الكرة ممتع، لكن دوري هنا هو...")
2. غياب المستوى الدراسي: إذا كان مستوى الطالب (غير محدد) وطلب استفساراً عن تسجيل المواد، اشرح له بأسلوب طبيعي أنك تحتاج لمعرفة مستواه من القائمة الجانبية لتتمكن من مساعدته بدقة.
3. منع الترشيحات: إذا طلب "نصيحة"، أو "أفضل مادة"، وضح له بمهنية أن دورك يقتصر على عرض بيانات اللائحة بموضوعية، وانصحه باستخدام "نظام الترشيحات المنفصل" المخصص لذلك.
4. الأماكن المحددة: "مدرج 1" و "مدرج 2" يقعان في الدور الثاني.
5. استخراج البيانات: اعتمد على "المعلومات الثابتة" واللائحة فقط. إذا لم تجد الإجابة، اعتذر بلطف وأخبره أن المعلومة غير متوفرة في الدليل الحالي بدلاً من الرد الآلي الجاف.
"""
                    user_template = f"""{hardcoded_knowledge}
المعلومات المرفقة من اللائحة:
{{context}}

تاريخ المحادثة:
{{chat_history}}

سؤال الطالب: {{input}}
إجابة المرشد الأكاديمي:"""

                else:
                    system_template = f"""You are a smart, polite academic advisor for Intelligent Systems Engineering.
{student_context}

Smart Refusal Rules (Maintain boundaries but sound natural):
1. Out of Scope: If asked about non-academic topics, briefly and politely acknowledge the topic to show understanding, then pivot to explaining that your focus is strictly on college academic advising.
2. Missing Level: If the student level is (غير محدد) and they ask about courses, naturally explain that you need them to select their level from the sidebar to give accurate information.
3. No Recommendations: If asked for the "best" course or advice, professionally explain that your role is to provide syllabus data objectively, and guide them to use the "separate recommendation system".
4. CRITICAL: "Hall 1" and "Hall 2" are located on the Second Floor.
5. ZERO HALLUCINATION: Rely ONLY on the hardcoded knowledge and syllabus. If information is missing, apologize naturally and state it's not in the current guide.
"""
                    user_template = f"""{hardcoded_knowledge}
Provided Context from Syllabus:
{{context}}

Chat History:
{{chat_history}}

Student's Question: {{input}}
Academic Advisor's Answer:"""

                prompt = ChatPromptTemplate.from_messages([
                    SystemMessagePromptTemplate.from_template(system_template),
                    HumanMessagePromptTemplate.from_template(user_template)
                ])

                docs = retriever.invoke(search_query)
                context_text = "\n\n".join([f"--- Page {doc.metadata.get('page', 0)} ---\n{doc.page_content}" for doc in docs])

                chain = prompt | llm | StrOutputParser()
                answer = chain.invoke({
                    "context": context_text,
                    "chat_history": chat_history,
                    "input": user_input
                })

            except Exception as e:
                answer = f"Error: {e}"

        render_message(answer)
    st.session_state.messages.append({"role": "assistant", "content": answer})
    st.rerun()
```
